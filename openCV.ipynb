```python
import cv2 as cv2
```

Image i/o

```python
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/test.jpg')
cv2_imshow(image)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
cv2.resize(gray, (200,200))
```

```python
blur = cv2.GaussianBlur(gray, (3,3), 0)
cv2_imshow(blur)
```

```python
edges = cv2.Canny(blur, 100, 200)
cv2_imshow(edges)
```

    #Drawing

```python
cv2.line(image, (0,0), (300,300), (255,0,0), 3)
cv2_imshow(image)
```

```python
cv2.rectangle(image, (0,0), (200,200), (0,255,0), 2)
cv2_imshow(image)
```

```python
cv2.circle(image, (100,100), 50, (0,0,255), -1)
cv2_imshow(image)
```

```python
cv2.putText(image, 'Scenary', (10,500), cv2.FONT_HERSHEY_SIMPLEX, 4, (255,255,255), 2)
cv2_imshow(image)
```

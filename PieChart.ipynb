```python
import matplotlib as plt
```

Pie Chart

```python
marks = [50,20,10,10,10]  #numerical coloumn
subject = ["DSA","DBMS","DAA","ML","VA"] #categorical coloumn
```

```python
plt.pyplot.pie(marks,labels=subject,autopct="%0.1f%%",colors=["grey",'pink','skyblue','yellow','brown'],explode=[.2,0.1,0.11,0.3,0.4])
plt.pyplot.show()
```

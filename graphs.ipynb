#BarGraph - (numerical vs categorical)

```python
marks=[40,50,33,56,88,12] #numerical
classs=['class-a','class-b','class-c','class-d','class-e','class-f'] #categorical
```

```python
plt.barh(classs,marks) #barh -> horizontal
plt.xlabel("Class")
plt.ylabel("Marks")
plt.title("Marks vs Classes")
```

```python
plt.bar(classs,marks) #bar-> vertical
plt.xlabel("Class")
plt.ylabel("Marks")
plt.title("Marks vs Classes")
```
